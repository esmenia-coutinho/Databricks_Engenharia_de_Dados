### ✅ Criação de Delta tables


```
# Função para dropar as tabelas no metastore
def dropar_tabelas(tabelas):
    for nome in tabelas.keys():
        spark.sql(f"DROP TABLE IF EXISTS lhdwePanex.{nome}")

# Dropar as tabelas no metastore
dropar_tabelas(tabelas)
```

```
# Caminho base dos dados
PATH_BASE = "dbfs:/mnt/panex/lhdw/gold/"

# Lista de tabelas e seus caminhos
tabelas = {
    "fato_vendas": f"{PATH_BASE}fato/",
    "produtos": f"{PATH_BASE}dimensao/produtos/",
    "categorias": f"{PATH_BASE}dimensao/categorias/",
    "vendedores": f"{PATH_BASE}dimensao/vendedores/",
    "clientes": f"{PATH_BASE}dimensao/clientes/",
    "cidades": f"{PATH_BASE}dimensao/cidades/",
    "paises": f"{PATH_BASE}dimensao/paises/"
}

# Criar o database lhdwePanex
spark.sql("CREATE DATABASE IF NOT EXISTS lhdwePanex")
spark.sql("USE lhdwePanex")

# Função para carregar as tabelas Delta e criar tabelas no metastore
def carregar_e_criar_tabelas(tabelas):
    dataframes = {}
    for nome, caminho in tabelas.items():
        df = spark.read.format("delta").load(caminho)
        dataframes[nome] = df
        df.createOrReplaceTempView(nome)
        spark.sql(f"CREATE TABLE IF NOT EXISTS lhdwePanex.{nome} USING DELTA LOCATION '{caminho}'")
    return dataframes

# Carregar as tabelas e criar tabelas no metastore
dfs = carregar_e_criar_tabelas(tabelas)

# Exibir os dataframes carregados
for nome, df in dfs.items():
    print(f"Tabela: {nome}")
    display(df)
```
